```python
import numpy as np
from scipy.stats import chisquare, chi2_contingency
```

```python
# Observed frequencies
observed = np.array([8, 12, 9, 11, 10, 10])

# Expected frequencies for a fair die
expected = np.array([10, 10, 10, 10, 10, 10])

# Chi-Square Goodness-of-Fit Test
chi_stat, p_value = chisquare(
    f_obs=observed,
    f_exp=expected
)

df = len(observed) - 1

print("=== Chi-Square Goodness-of-Fit Test ===")
print("Chi-Square Statistic:", chi_stat)
print("Degrees of Freedom:", df)
print("p-value:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Decision: Reject the null hypothesis.")
    print("Conclusion: The die is not fair.")
else:
    print("Decision: Fail to reject the null hypothesis.")
    print("Conclusion: There is no significant evidence that the die is unfair.")
```

```python
# Contingency table
table = np.array([
    [30, 40, 30],   # Male
    [40, 30, 30]    # Female
])

# Chi-Square Test of Independence
chi_stat, p_value, df, expected = chi2_contingency(table)

print("=== Chi-Square Test of Independence ===")
print("Chi-Square Statistic:", chi_stat)
print("Degrees of Freedom:", df)
print("p-value:", p_value)

print("\nExpected Frequencies:")
print(expected)
```

```python
%%R

observed <- c(8, 12, 9, 11, 10, 10)

expected_prob <- rep(1/6, 6)

result1 <- chisq.test(
  observed,
  p = expected_prob
)

print(result1)
```

```python
%%R

table_data <- matrix(
  c(30, 40, 30,
    40, 30, 30),
  nrow = 2,
  byrow = TRUE
)

rownames(table_data) <- c("Male", "Female")
colnames(table_data) <- c(
  "Product A",
  "Product B",
  "Product C"
)

result2 <- chisq.test(table_data)

print(result2)
```
